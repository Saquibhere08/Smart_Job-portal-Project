# Basics of Probability

## Question 1a — Coin Toss Simulation (10,000 trials)

When we toss a fair coin, the theoretical probability of getting heads or tails is 0.5 each. However, in practice, we can only observe an *experimental* (empirical) probability by actually performing the experiment many times. The **Law of Large Numbers** tells us that as the number of trials increases, the experimental probability converges towards the theoretical probability.

Here, we simulate 10,000 coin tosses using Python's `random` module, count how many times heads and tails appear, and then compute the experimental probability for each outcome.

```python
# 1a. Tossing a coin 10,000 times
num_tosses = 10000
heads_count = 0
tails_count = 0

for _ in range(num_tosses):
    toss = random.choice(['Heads', 'Tails'])
    if toss == 'Heads':
        heads_count += 1
    else:
        tails_count += 1

prob_heads = heads_count / num_tosses
prob_tails = tails_count / num_tosses

print(f"Total Tosses: {num_tosses}")
print(f"Heads: {heads_count} | Experimental Probability: {prob_heads:.4f}")
print(f"Tails: {tails_count} | Experimental Probability: {prob_tails:.4f}")
print(f"\nTheoretical Probability: 0.5000 each")
```

**Explanation:** We loop 10,000 times, randomly picking either 'Heads' or 'Tails' in each iteration. After the loop, we divide the count of each outcome by the total number of tosses to get the experimental probability. As expected, both probabilities are very close to 0.5, which matches the theoretical value for a fair coin.

## Question 1b — Rolling Two Dice: Probability of Sum = 7

When rolling two fair dice, there are 36 possible outcomes (6 × 6). Out of these, 6 combinations give a sum of 7: (1,6), (2,5), (3,4), (4,3), (5,2), (6,1). So the theoretical probability is 6/36 = 1/6 ≈ 0.1667.

We simulate rolling two dice many times and count how often the sum equals 7 to estimate this probability experimentally.

```python
# 1b. Rolling two dice and computing probability of sum = 7
num_rolls = 10000
sum_seven_count = 0

for _ in range(num_rolls):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)
    if die1 + die2 == 7:
        sum_seven_count += 1

prob_sum_seven = sum_seven_count / num_rolls

print(f"Total Rolls: {num_rolls}")
print(f"Times Sum = 7: {sum_seven_count}")
print(f"Experimental Probability of Sum = 7: {prob_sum_seven:.4f}")
print(f"Theoretical Probability of Sum = 7:  {6/36:.4f}")
```

**Explanation:** In each trial we roll two dice (each generating a random integer from 1 to 6) and check if their sum is 7. After 10,000 trials, the experimental probability is computed by dividing the number of successful outcomes by the total number of trials. The result closely approximates the theoretical probability of 1/6.

## Question 2 — Probability of At Least One "6" in 10 Rolls

The probability of NOT getting a 6 in a single roll is 5/6. The probability of NOT getting any 6 in 10 independent rolls is (5/6)^10. Therefore, the probability of getting **at least one** 6 is:

$$P(\text{at least one 6}) = 1 - \left(\frac{5}{6}\right)^{10} \approx 0.8385$$

We write a function that estimates this through simulation by repeating the experiment many times and checking how often at least one 6 appears.

```python
def estimate_at_least_one_six(num_rolls=10, num_trials=100000):
    """
    Estimates the probability of getting at least one '6'
    in a given number of rolls of a fair die.
    """
    success_count = 0

    for _ in range(num_trials):
        # Simulate rolling a die 'num_rolls' times
        got_six = False
        for _ in range(num_rolls):
            if random.randint(1, 6) == 6:
                got_six = True
                break  # No need to roll further
        if got_six:
            success_count += 1

    return success_count / num_trials

# Run the estimation
estimated_prob = estimate_at_least_one_six()
theoretical_prob = 1 - (5/6)**10

print(f"Estimated Probability of at least one '6' in 10 rolls: {estimated_prob:.4f}")
print(f"Theoretical Probability: {theoretical_prob:.4f}")
```

**Explanation:** The function runs 100,000 independent trials. In each trial, we roll a die 10 times and check if any roll results in a 6. If yes, we count it as a success. The proportion of successful trials gives us the estimated probability. The result closely matches the theoretical value of approximately 0.8385.

# Conditional Probability and Bayes' Theorem

## Question 3 — Ball Drawing Simulation with Replacement

A bag contains 5 red, 7 green, and 8 blue balls (total = 20). Since balls are drawn **with replacement**, each draw is independent of the previous one. This means:
- P(Red) = 5/20 = 0.25
- P(Green) = 7/20 = 0.35
- P(Blue) = 8/20 = 0.40

### Part (a): P(Red | Previous was Blue)
Since the draws are with replacement, they are independent. So theoretically, P(Red | Previous Blue) = P(Red) = 0.25. We verify this by simulation.

### Part (b): Verifying Bayes' Theorem
Bayes' Theorem states:

$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

We verify: $P(\text{Red}|\text{Blue}) = \frac{P(\text{Blue}|\text{Red}) \cdot P(\text{Red})}{P(\text{Blue})}$

```python
# Question 3 - Ball Drawing Simulation
num_draws = 1000
balls = ['Red'] * 5 + ['Green'] * 7 + ['Blue'] * 8  # 20 balls total

# Simulate 1000 draws with replacement
draws = [random.choice(balls) for _ in range(num_draws)]

# --- Part (a): P(Red | Previous was Blue) ---
# Count consecutive pairs where previous = Blue and current = Red
blue_then_red = 0
blue_previous = 0

for i in range(1, num_draws):
    if draws[i - 1] == 'Blue':
        blue_previous += 1
        if draws[i] == 'Red':
            blue_then_red += 1

cond_prob_red_given_blue = blue_then_red / blue_previous if blue_previous > 0 else 0

print("=== Part (a): P(Red | Previous was Blue) ===")
print(f"Times previous draw was Blue: {blue_previous}")
print(f"Times Blue was followed by Red: {blue_then_red}")
print(f"Estimated P(Red | Previous Blue): {cond_prob_red_given_blue:.4f}")
print(f"Theoretical P(Red) = 5/20 = {5/20:.4f}")
print(f"(Since draws are with replacement, events are independent)")
```

```python
# --- Part (b): Verify Bayes' Theorem ---
# Using simulation data to verify:
# P(Red | Blue) = P(Blue | Red) * P(Red) / P(Blue)

# Calculate P(Blue | Previous was Red)
red_then_blue = 0
red_previous = 0

for i in range(1, num_draws):
    if draws[i - 1] == 'Red':
        red_previous += 1
        if draws[i] == 'Blue':
            red_then_blue += 1

cond_prob_blue_given_red = red_then_blue / red_previous if red_previous > 0 else 0

# Marginal probabilities from simulation
p_red_sim = draws.count('Red') / num_draws
p_blue_sim = draws.count('Blue') / num_draws

# Bayes' Theorem: P(Red | Blue) = P(Blue | Red) * P(Red) / P(Blue)
bayes_result = (cond_prob_blue_given_red * p_red_sim) / p_blue_sim if p_blue_sim > 0 else 0

print("=== Part (b): Verifying Bayes' Theorem ===")
print(f"\nSimulation-based marginal probabilities:")
print(f"  P(Red)  = {p_red_sim:.4f}")
print(f"  P(Blue) = {p_blue_sim:.4f}")
print(f"\nConditional probabilities from simulation:")
print(f"  P(Blue | Previous Red) = {cond_prob_blue_given_red:.4f}")
print(f"  P(Red | Previous Blue) = {cond_prob_red_given_blue:.4f}  [directly observed]")
print(f"\nBayes' Theorem: P(Red|Blue) = P(Blue|Red) * P(Red) / P(Blue)")
print(f"  Bayes' Result = {cond_prob_blue_given_red:.4f} * {p_red_sim:.4f} / {p_blue_sim:.4f} = {bayes_result:.4f}")
print(f"  Direct Observation = {cond_prob_red_given_blue:.4f}")
print(f"\n✓ Both values are approximately equal, verifying Bayes' Theorem!")
```

**Explanation:** We simulate 1000 draws with replacement. For part (a), we look at all consecutive pairs where the previous draw was Blue and check how many times the current draw was Red. Since draws are independent (with replacement), P(Red | Previous Blue) ≈ P(Red) = 0.25.

For part (b), we compute P(Blue | Previous Red) from the data and then apply Bayes' formula. The result from Bayes' theorem matches the directly observed conditional probability, confirming the theorem holds.

# Random Variables and Discrete Probability

## Question 4 — Discrete Random Variable Simulation

A discrete random variable X can take values {1, 2, 3} with probabilities:
- P(X=1) = 0.25
- P(X=2) = 0.35
- P(X=3) = 0.40

We use `numpy.random.choice()` to generate 1000 samples according to this distribution, and then compute the empirical (sample) mean, variance, and standard deviation.

The **theoretical values** are:
- Mean: $E[X] = 1(0.25) + 2(0.35) + 3(0.40) = 2.15$
- Variance: $Var(X) = E[X^2] - (E[X])^2 = (1^2 \cdot 0.25 + 2^2 \cdot 0.35 + 3^2 \cdot 0.40) - 2.15^2 = 5.25 - 4.6225 = 0.6275$
- Std Dev: $\sigma = \sqrt{0.6275} \approx 0.7922$

```python
# Question 4 - Discrete Random Variable
values = [1, 2, 3]
probabilities = [0.25, 0.35, 0.40]

# Generate a sample of size 1000
sample = np.random.choice(values, size=1000, p=probabilities)

# Compute empirical statistics
empirical_mean = np.mean(sample)
empirical_variance = np.var(sample)
empirical_std = np.std(sample)

# Theoretical values
theoretical_mean = 1*0.25 + 2*0.35 + 3*0.40
theoretical_variance = (1**2 * 0.25 + 2**2 * 0.35 + 3**2 * 0.40) - theoretical_mean**2
theoretical_std = np.sqrt(theoretical_variance)

print("=== Discrete Random Variable (1000 samples) ===")
print(f"{'Statistic':<20} {'Empirical':<15} {'Theoretical':<15}")
print("-" * 50)
print(f"{'Mean':<20} {empirical_mean:<15.4f} {theoretical_mean:<15.4f}")
print(f"{'Variance':<20} {empirical_variance:<15.4f} {theoretical_variance:<15.4f}")
print(f"{'Std Deviation':<20} {empirical_std:<15.4f} {theoretical_std:<15.4f}")

# Show frequency distribution
print(f"\nSample Frequency Distribution:")
for v in values:
    count = np.sum(sample == v)
    print(f"  X={v}: {count} times ({count/1000:.3f}) | Expected: {probabilities[v-1]:.3f}")
```

**Explanation:** We use `np.random.choice()` with the `p` parameter to sample from a custom discrete distribution. The function randomly picks values from {1, 2, 3} according to the specified probabilities. We then use NumPy's `mean()`, `var()`, and `std()` functions to compute the sample statistics. The empirical values closely match the theoretical calculations, demonstrating that the simulation correctly follows the given distribution.

# Continuous Random Variables

## Question 5 — Exponential Distribution Simulation

The **exponential distribution** models the time between events in a Poisson process. It has a single parameter — the mean (or scale). If the mean is 5, the rate parameter λ = 1/5 = 0.2.

The PDF of the exponential distribution is:

$$f(x) = \lambda e^{-\lambda x}, \quad x \geq 0$$

We generate 2000 random samples and visualize them with:
- A **histogram** showing the frequency distribution of the samples
- A **PDF overlay** showing the theoretical curve on top of the histogram

```python
# Question 5 - Exponential Distribution
mean_val = 5  # Mean of the exponential distribution
sample_size = 2000

# Generate random samples from exponential distribution
exp_samples = np.random.exponential(scale=mean_val, size=sample_size)

# Create the figure
plt.figure(figsize=(10, 6))

# 5a. Histogram
plt.hist(exp_samples, bins=40, density=True, alpha=0.7, color='#42A5F5',
         edgecolor='black', label='Histogram (Simulated Data)')

# 5b. PDF Overlay
x_vals = np.linspace(0, max(exp_samples), 500)
lambda_val = 1 / mean_val
pdf_vals = lambda_val * np.exp(-lambda_val * x_vals)
plt.plot(x_vals, pdf_vals, 'r-', linewidth=2.5, label=f'Theoretical PDF (λ = {lambda_val})')

plt.title('Exponential Distribution (Mean = 5)', fontsize=16)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Probability Density', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Sample Mean: {np.mean(exp_samples):.4f} (Theoretical: {mean_val})")
print(f"Sample Std Dev: {np.std(exp_samples):.4f} (Theoretical: {mean_val})")
```

**Explanation:** We use `np.random.exponential()` with `scale=5` to generate 2000 samples. The histogram (with `density=True` for normalization) shows the empirical distribution of the data. We overlay the theoretical PDF $f(x) = 0.2 \cdot e^{-0.2x}$ as a red curve. The histogram closely follows the PDF curve, confirming that our simulated data matches the theoretical exponential distribution. For an exponential distribution, the mean and standard deviation are both equal to the scale parameter (5).

# Central Limit Theorem

## Question 6 — Simulating the Central Limit Theorem (CLT)

The **Central Limit Theorem** states that the distribution of sample means approaches a normal (bell-shaped) distribution as the sample size increases, regardless of the shape of the original population distribution.

To demonstrate this:
1. We start with a **uniform distribution** (which is flat, not bell-shaped).
2. We draw 1000 samples of size n=30 from this distribution.
3. We compute the mean of each sample.
4. We plot the distribution of these 1000 sample means — it should look approximately normal.

This is a powerful result because it tells us that sample means are normally distributed even when the underlying data is not.

```python
# Question 6 - Central Limit Theorem

# Step (a): Generate 10,000 random numbers from a uniform distribution
population = np.random.uniform(low=0, high=1, size=10000)

# Step (b): Draw 1000 samples of size n=30 and calculate their means
num_samples = 1000
sample_size = 30
sample_means = []

for _ in range(num_samples):
    sample = np.random.choice(population, size=sample_size, replace=True)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

# Step (c): Visualize both distributions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Original Uniform Distribution
axes[0].hist(population, bins=40, color='#FF9800', edgecolor='black', alpha=0.8)
axes[0].set_title('Original Uniform Distribution\n(Population: 10,000 values)', fontsize=14)
axes[0].set_xlabel('Value', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].grid(True, alpha=0.3)

# Plot 2: Distribution of Sample Means
axes[1].hist(sample_means, bins=40, color='#4CAF50', edgecolor='black', alpha=0.8)
axes[1].set_title('Distribution of Sample Means\n(1000 samples, n=30 each)', fontsize=14)
axes[1].set_xlabel('Sample Mean', fontsize=12)
axes[1].set_ylabel('Frequency', fontsize=12)
axes[1].axvline(x=np.mean(sample_means), color='red', linestyle='--', linewidth=2,
                label=f'Grand Mean = {np.mean(sample_means):.4f}')
axes[1].legend(fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Population Mean: {np.mean(population):.4f}")
print(f"Mean of Sample Means: {np.mean(sample_means):.4f}")
print(f"Std Dev of Sample Means: {np.std(sample_means):.4f}")
print(f"Theoretical Std Dev of Sample Means (σ/√n): {np.std(population)/np.sqrt(sample_size):.4f}")
```

**Explanation:** The left plot shows the original uniform distribution — it's flat and spread evenly between 0 and 1. The right plot shows the distribution of 1000 sample means (each computed from a random sample of 30 values). Despite the original data being uniform (non-normal), the sample means form a clear **bell-shaped (normal) curve**, perfectly illustrating the Central Limit Theorem.

Key observations:
- The mean of the sample means is approximately equal to the population mean (~0.5).
- The standard deviation of the sample means is approximately $\sigma / \sqrt{n}$, where $\sigma$ is the population standard deviation and $n=30$ is the sample size.
- This demonstrates that regardless of the population distribution, the sampling distribution of the mean tends toward a normal distribution as sample size increases.
