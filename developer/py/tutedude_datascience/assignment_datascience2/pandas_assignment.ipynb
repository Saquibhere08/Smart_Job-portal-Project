## Step 2 - Import Pandas

```python
import pandas as pd
```

---
## 1. Create a DataFrame

A DataFrame is a two-dimensional, tabular data structure in Pandas. We create one from a dictionary containing employee information such as name, age, department, and salary.

```python
# Create a DataFrame with the given data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'HR'],
    'Salary': [45000, 54000, 50000, 62000, 47000]
}

df = pd.DataFrame(data)
print("DataFrame created successfully!")
```

### 1a. Print the first five rows

The `head()` method displays the first N rows of the DataFrame. By default, it shows the first 5 rows.

```python
# Print the first five rows of the DataFrame
print("First five rows of the DataFrame:")
df.head()
```

### 1b. Summary statistics of 'Age' and 'Salary'

The `describe()` method provides summary statistics including count, mean, standard deviation, min, max, and quartiles for numerical columns.

```python
# Get summary statistics of 'Age' and 'Salary' columns
print("Summary Statistics for 'Age' and 'Salary':")
df[['Age', 'Salary']].describe()
```

### 1c. Average salary of employees in the 'HR' department

We filter the DataFrame to include only rows where the Department is 'HR', and then calculate the mean of the 'Salary' column.

```python
# Calculate the average salary of employees in the 'HR' department
hr_avg_salary = df[df['Department'] == 'HR']['Salary'].mean()
print(f"Average Salary in HR Department: {hr_avg_salary}")
```

---
## 2. Add a 'Bonus' Column

We add a new column called 'Bonus' which is calculated as 10% of each employee's salary. This demonstrates how to create derived columns in a DataFrame.

```python
# Add a new column 'Bonus' which is 10% of the salary
df['Bonus'] = df['Salary'] * 0.10
print("DataFrame with Bonus column:")
df
```

---
## 3. Filter Employees Aged Between 25 and 30

Filtering allows us to select rows that meet specific conditions. Here we filter employees whose age falls between 25 and 30 (inclusive).

```python
# Filter employees aged between 25 and 30
filtered_df = df[(df['Age'] >= 25) & (df['Age'] <= 30)]
print("Employees aged between 25 and 30:")
filtered_df
```

---
## 4. Group by Department and Calculate Average Salary

The `groupby()` method groups data by a specific column. We group by 'Department' and calculate the average salary for each department to understand salary distribution across the organization.

```python
# Group by 'Department' and calculate the average salary
dept_avg_salary = df.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
dept_avg_salary
```

---
## 5. Sort by Salary and Save to CSV

Sorting organizes data in a specific order. We sort the DataFrame by 'Salary' in ascending order and save the result to a CSV file using the `to_csv()` method. The `index=False` parameter prevents writing the row indices to the file.

```python
# Sort the DataFrame by 'Salary' in ascending order
sorted_df = df.sort_values(by='Salary', ascending=True)
print("DataFrame sorted by Salary (ascending):")
print(sorted_df)

# Save the sorted DataFrame to a CSV file
sorted_df.to_csv('sorted_employees.csv', index=False)
print("\nSorted DataFrame saved to 'sorted_employees.csv'")
```
