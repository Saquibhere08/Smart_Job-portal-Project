## 1. 1D Array Operations

We create a 1D NumPy array with integers from 1 to 20 and perform basic statistical operations on it. These operations help us understand the distribution and characteristics of the data.

```python
# Create a 1D NumPy array with integers from 1 to 20
arr_1d = np.arange(1, 21)
print("1D Array:", arr_1d)
```

### 1a. Calculate sum, mean, median, and standard deviation

These are fundamental statistical measures:
- **Sum** gives the total of all elements.
- **Mean** gives the average value.
- **Median** gives the middle value when sorted.
- **Standard Deviation** measures the spread of the data.

```python
# Calculate sum, mean, median, and standard deviation
print("Sum:", np.sum(arr_1d))
print("Mean:", np.mean(arr_1d))
print("Median:", np.median(arr_1d))
print("Standard Deviation:", np.std(arr_1d))
```

### 1b. Find indices of elements greater than 10

We use `np.where()` to find the indices where a given condition is true. This is useful for filtering data based on criteria.

```python
# Find the indices of elements greater than 10
indices = np.where(arr_1d > 10)
print("Indices of elements greater than 10:", indices[0])
print("Elements greater than 10:", arr_1d[indices])
```

## 2. 2D Array Operations

We create a 2D array of shape 4×4 with numbers from 1 to 16. 2D arrays are used to represent matrices, which are fundamental in linear algebra and data science.

```python
# Create a 2D NumPy array of shape 4x4 with numbers from 1 to 16
arr_2d = np.arange(1, 17).reshape(4, 4)
print("2D Array (4x4):")
print(arr_2d)
```

### 2b. Transpose of the array

The transpose of a matrix swaps its rows and columns. This is a common operation in linear algebra.

```python
# Find the transpose of the array
print("Transpose of the array:")
print(arr_2d.T)
```

### 2c. Row-wise and column-wise sums

- `axis=1` computes the sum across each row.
- `axis=0` computes the sum down each column.

```python
# Calculate row-wise and column-wise sums
print("Row-wise sums:", np.sum(arr_2d, axis=1))
print("Column-wise sums:", np.sum(arr_2d, axis=0))
```

## 3. Operations on Two 3×3 Arrays

We create two 3×3 arrays filled with random integers between 1 and 20, then perform element-wise arithmetic and the dot product. The dot product is a key operation in matrix multiplication.

```python
# Create two 3x3 arrays with random integers between 1 and 20
np.random.seed(42)  # Setting seed for reproducibility
array_a = np.random.randint(1, 21, size=(3, 3))
array_b = np.random.randint(1, 21, size=(3, 3))

print("Array A:")
print(array_a)
print("\nArray B:")
print(array_b)
```

### 3a. Element-wise addition, subtraction, and multiplication

Element-wise operations are performed on corresponding elements of the two arrays.

```python
# Element-wise addition
print("Element-wise Addition:")
print(array_a + array_b)

# Element-wise subtraction
print("\nElement-wise Subtraction:")
print(array_a - array_b)

# Element-wise multiplication
print("\nElement-wise Multiplication:")
print(array_a * array_b)
```

### 3b. Dot product

The dot product (matrix multiplication) computes the sum of products of corresponding entries in the rows and columns of the two matrices.

```python
# Dot product of the two arrays
print("Dot Product:")
print(np.dot(array_a, array_b))
```

## 4. Reshape and Slice

Reshaping allows us to change the dimensions of an array without changing its data. Slicing lets us extract specific portions of the array. Here we reshape a 1D array of 12 elements into a 3×4 matrix and then slice the first two rows and last two columns.

```python
# Create a 1D array of size 12
arr_12 = np.arange(1, 13)
print("Original 1D Array:", arr_12)

# Reshape into a 3x4 2D array
arr_reshaped = arr_12.reshape(3, 4)
print("\nReshaped 3x4 Array:")
print(arr_reshaped)

# Slice: first two rows and last two columns
sliced = arr_reshaped[:2, -2:]
print("\nSliced (first 2 rows, last 2 columns):")
print(sliced)
```
