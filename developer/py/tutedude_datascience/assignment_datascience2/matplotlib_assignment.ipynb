## Step 2 - Import Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

---
## 1. Simple Line Plot

A line plot is used to display data points connected by straight line segments. It is commonly used to visualize trends over time or continuous data. Here we plot a simple dataset and customize it with a title, axis labels, and a grid for better readability.

```python
# Data for line plot
x = [1, 2, 3, 4, 5]
y = [10, 15, 25, 30, 50]

# 1a. Plot the data
plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', color='blue', linestyle='-', linewidth=2, markersize=8)

# 1b. Customize the plot with title, axis labels, and grid
plt.title('Simple Line Plot', fontsize=16)
plt.xlabel('X-Axis', fontsize=12)
plt.ylabel('Y-Axis', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

---
## 2. Bar Graph - Student Marks

A bar graph is used to compare quantities across categories. It displays rectangular bars with lengths proportional to the values they represent. Here we visualize the marks scored by students in a subject, with customized colors and a title.

```python
# Data for bar graph
students = ['John', 'Jane', 'Alice', 'Bob']
marks = [75, 85, 60, 90]

# 2a. Plot the data as a bar graph
plt.figure(figsize=(8, 5))
colors = ['#4CAF50', '#2196F3', '#FF9800', '#F44336']
bars = plt.bar(students, marks, color=colors, edgecolor='black', width=0.6)

# 2b. Customize colors and add a title
plt.title('Marks Scored by Students', fontsize=16)
plt.xlabel('Students', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)

# Add value labels on top of each bar
for bar, mark in zip(bars, marks):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 2,
             str(mark), ha='center', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

---
## 3. Pie Chart - Company Revenue by Region

A pie chart is used to show the proportional distribution of categories in a dataset. Each slice of the pie represents a category's contribution to the total. We highlight (explode) the region with the highest revenue — North America — to draw attention to it.

```python
# Data for pie chart
regions = ['North America', 'Europe', 'Asia', 'Others']
revenue = [45, 25, 20, 10]

# Highlight the region with the highest revenue (North America)
explode = (0.1, 0, 0, 0)  # Explode the first slice

# 3a & 3b. Create pie chart with labels and highlight
plt.figure(figsize=(8, 6))
colors = ['#FF6384', '#36A2EB', '#FFCE56', '#4BC0C0']
plt.pie(revenue, labels=regions, autopct='%1.1f%%', startangle=140,
        colors=colors, explode=explode, shadow=True, textprops={'fontsize': 12})

plt.title('Company Revenue Distribution by Region', fontsize=16)
plt.axis('equal')  # Equal aspect ratio ensures the pie is circular

plt.tight_layout()
plt.show()
```

---
## 4. Histogram - Frequency Distribution

A histogram is used to visualize the frequency distribution of a dataset. It divides data into bins (intervals) and shows how many data points fall into each bin. Here we generate 1000 random integers between 1 and 100 and plot their distribution.

```python
# Generate 1000 random integers between 1 and 100
np.random.seed(42)  # For reproducibility
random_data = np.random.randint(1, 101, size=1000)

# Create a histogram
plt.figure(figsize=(10, 6))
plt.hist(random_data, bins=20, color='#6C63FF', edgecolor='black', alpha=0.8)

plt.title('Frequency Distribution of Random Integers (1-100)', fontsize=16)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
